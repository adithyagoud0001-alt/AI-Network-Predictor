# AI-Based Network Connection Predictor: Machine Learning Model Benchmark

This notebook trains and compares 5 candidate models:
1. Logistic Regression
2. Decision Tree
3. Random Forest
4. Gradient Boosting
5. XGBoost

Evaluated using 5-Fold Stratified Cross-Validation, Macro F1, Precision, Recall, and Confusion Matrices on independent held-out test data.

```python
import sys
from pathlib import Path
sys.path.append(str(Path('../backend').resolve()))

import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load model training metadata
meta_path = Path('../backend/app/ml/models/model_metadata.json')
with open(meta_path, 'r') as f:
    meta = json.load(f)

print(f"Champion Model Selected: {meta['champion_model_name']}")
print(f"Held-Out Test Accuracy: {meta['champion_test_metrics']['accuracy'] * 100:.2f}%")
print(f"Held-Out Test Macro F1: {meta['champion_test_metrics']['macro_f1'] * 100:.2f}%")
```

```python
# Algorithm Comparison Table
comp_df = pd.DataFrame(meta['validation_comparison'])
comp_df[['model_name', 'cv_macro_f1_mean', 'val_accuracy', 'val_macro_f1', 'val_macro_precision', 'val_macro_recall', 'train_duration_sec']]
```

```python
# Confusion Matrix Visualization
cm = meta['champion_test_metrics']['confusion_matrix']
classes = meta['classes']

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', xticklabels=classes, yticklabels=classes)
plt.title(f"Test Confusion Matrix - {meta['champion_model_name']}")
plt.xlabel('Predicted Class')
plt.ylabel('Actual Class')
plt.show()
```

```python
# Top Global Feature Importances
feat_df = pd.DataFrame(meta['global_feature_importances'])
plt.figure(figsize=(8, 4))
sns.barplot(data=feat_df.head(6), x='importance_pct', y='feature', palette='viridis')
plt.title('Top 6 Contributing Feature Weights (%)')
plt.xlabel('Importance Percentage')
plt.show()
```
