# AI-Based Network Connection Predictor: Exploratory Data Analysis (EDA)

This notebook performs exploratory analysis on collected and generated network telemetry:
- Missing value audits (including verifying null handling on wired Ethernet interfaces)
- QoS metric statistical distributions
- Outlier detection using Interquartile Range (IQR)
- Correlation matrices across network parameters
- Class balance analysis (GOOD, MODERATE, POOR)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load dataset
dataset_path = Path('../backend/app/data/datasets/network_telemetry_dataset.csv')
df = pd.read_csv(dataset_path)
print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# Class Balance Evaluation
print("Class Distribution:")
print(df['heuristic_quality_label'].value_counts(normalize=True) * 100)

plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='heuristic_quality_label', palette=['#10b981', '#ef4444', '#f59e0b'])
plt.title('Distribution of Network Quality Classes')
plt.xlabel('Quality Class')
plt.ylabel('Count')
plt.show()
```

```python
# Missing value audit across connection types
print("Missing Values per Column:")
print(df.isnull().sum())

# Validate Ethernet signal null handling
eth_rows = df[df['connection_type'] == 'Ethernet']
print(f"\nEthernet rows with Null RSSI: {eth_rows['signal_strength'].isnull().sum()} / {len(eth_rows)} (100% Null expected)")
```

```python
# Correlation Heatmap
num_cols = ['latency', 'packet_loss', 'jitter', 'download_bandwidth', 'upload_bandwidth', 'network_utilization']
plt.figure(figsize=(8, 6))
sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Network Parameter Correlation Heatmap')
plt.show()
```
